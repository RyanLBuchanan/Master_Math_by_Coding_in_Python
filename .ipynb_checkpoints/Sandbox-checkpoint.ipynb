```python
import sympy as sym
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display, Math
```

```python
x = sym.symbols("x")

fx = x**3

lim_pnt = 1.5

lim = sym.limit(fx, x, lim_pnt)

#display(Math(f"lim = {lim}"))

display(Math('\\lim_{x\\to %g} %s = %g' %(lim_pnt, sym.latex(fx), lim)))
```
